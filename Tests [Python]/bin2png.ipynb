Converting images to .bin file

```python
import os
import numpy as np
from PIL import Image

def bin_to_png_all(processed_dir="processed", output_dir="png_images"):
    """
    Converts all .bin grayscale images from processed_dir to .png images.
    Expects filenames in the format: name_WIDTHxHEIGHT_rest.bin
    
    Example: lena_512x512_bypass_hw.bin -> 512x512 image
    """
    os.makedirs(output_dir, exist_ok=True)
    
    for file_name in os.listdir(processed_dir):
        if not file_name.lower().endswith(".bin"):
            continue
        
        # Parse dimensions from filename
        parts = file_name.split("_")
        if len(parts) < 2 or "x" not in parts[1]:
            print(f"Skipping {file_name} (no dimensions found)")
            continue
        
        try:
            w, h = map(int, parts[1].split("x"))
        except ValueError:
            print(f"Skipping {file_name} (invalid dimensions)")
            continue
        
        file_path = os.path.join(processed_dir, file_name)
        
        # Read binary data
        with open(file_path, "rb") as f:
            data = f.read()
        
        expected_size = w * h
        if len(data) != expected_size:
            print(f"Skipping {file_name} (size mismatch: {len(data)} vs {expected_size})")
            continue
        
        img = np.frombuffer(data, dtype=np.uint8).reshape((h, w))
        
        # Save as PNG
        out_name = os.path.splitext(file_name)[0] + ".png"
        out_path = os.path.join(output_dir, out_name)
        Image.fromarray(img).save(out_path)
        print(f"Converted {file_name} → {out_name}")
```

```python
bin_to_png_all()
```
