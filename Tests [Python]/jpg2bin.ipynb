## Converting images to .bin file

```python
from PIL import Image
import numpy as np
import os

def jpg_to_bin_grayscale(file_name: str, max_size: int = 512):
    """
    Converts a grayscale JPG from raw_images/ to a .bin in images/.
    
    Output file is named: fileName_WIDTHxHEIGHT.bin
    """
    input_path = os.path.join("raw_images", file_name)
    
    # Open and convert to grayscale
    img = Image.open(input_path).convert("L")
    
    # Resize if needed
    w, h = img.size
    if max(w, h) > max_size:
        scale = max_size / max(w, h)
        new_size = (int(w * scale), int(h * scale))
        img = img.resize(new_size, Image.Resampling.LANCZOS)
        print(f"Resized {file_name} → {new_size}")
    else:
        new_size = (w, h)
        print(f"No resize needed for {file_name}, size={new_size}")
    
    # Create output file name
    base_name, _ = os.path.splitext(file_name)
    out_name = f"{base_name}_{new_size[0]}x{new_size[1]}.bin"
    output_path = os.path.join("images", out_name)
    
    # Save as raw binary
    arr = np.array(img, dtype=np.uint8)
    arr.tofile(output_path)
    
    print(f"Saved {output_path} ({arr.shape[1]}x{arr.shape[0]})")
```

### Lena (512x512)

```python
jpg_to_bin_grayscale("lena.jpg")
```
